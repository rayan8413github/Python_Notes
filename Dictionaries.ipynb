# Dictionaries - The dictionary is a special type of data structure in python which stores data in form of key:value pair, where a dictionary can have duplicate values but not duplicate keys (or we say that keys can't be repeated) and must be immutable. Values can be of any datatype in dictionary. 

# Methods to create Dictionaries

```python
d = {'A':21, 'B':31, 'C':14, 'D':91}
print(d)
```

```python
# using dict() constructor to make dictionary
f = dict(a = 'Ram', b = 13, c = True, d = "Japan")
print(f)
```

# Note - Dictionaries are case sensitive

# Accessing Dictionary elements

```python
d = {'a':21, 'b':6.2, 'c':True, 'd':"Pen"}
print(d['a'])
print(d.get('a'))
```

# Adding and updating new items in dictionary

```python
a1 = {'a':43, 'b':"Laptop", 'c':2.9}
a1['d'] = True  #adding new element
a1['b'] = "Charger" #updating existing element
a1
```
