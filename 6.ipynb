```python
b = int(input("Enter a number :"))
if b%2 == 0:
    print("even number.")
```

# Ternary Operator or Conditional Operator in Python

```python
a = int(input("Enter a number:"))
print("a is 5." if a == 5 else "a is not 5.")
```

# Looping Statements

```python
for i in range(5):
    print(i)
```

```python
count = 0
while count < 5:
    print(count)
    count = count + 1
```

# Jump Statements

```python
# Break Statements - it is used to primarily leave a loop permanently.
for i in range(5):
    if i == 3:
        break
    print(i,end = "")
```

```python
# continue - it is used to skip an iteration(s)
for i in range(9):
    if i % 2 == 0:
        continue
    print(i)
```
