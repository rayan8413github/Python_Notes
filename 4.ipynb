# Tuples - are immutable ordered collection of elements. They are similar to lists but the key difference is that while lists are mutable tuples aren't.

# creating tuples

```python
# creating an empty tuple
tup = ()
print(tup)
```

```python
# built-in functions of tuples in python
tup1 = ("Aditya")
print(tup1)
print(tup1[0])
```

```python
tup1 = ("rat", 43, 6.9, False)
tup1
```

```python
#creating nested tuples
tup1 = (3, 6)
tup2 = ("monday", 2.0)
tup3 = (tup1, tup2)
tup3
```

# using for loop to create tuples

```python
tup = ("March", 5.9)
n = 3
for i in range(n):
    tup = (tup,)
    print(tup)
```

# Accesing elements, indexing and slicing of a tuple are same as list and strings

```python
n = "Rayan"
tup = tuple(n)
tup[1:4]
```

# Tuple Unpacking

```python
tup = ("Python", 5.1, False)
a,b,c = tup
print(a)
print(b)
print(c)
```

# Concatenation of Tuples

```python
a = ("p", "q", "r", "s")
b = (3,4,5,6)
c = a + b
print(c)
```

```python
# Tuples are also having negative indices like lists and strings
```

# Deleting a Tuple

```python
t1 = (3, 6, 2, 8)
del t1
#print(t1) gives an error now
```

```python
# making a tuple for another iterable
b = ["You", 65, 8.9, False]
a = ([x for x in b], 98)
print(a)
```

# Interchanging of List and Tuple

```python
t2 = (9, "Notebook", True, [7,"A", 9.6])
a = list(t2)
a.append("owl")
t2 = tuple(a)
print(t2)
```
