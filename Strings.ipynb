# String - It is  a datatype in python. It is a sequence of characters enclosed in quotes. We can primarily write it inside single/double/triple quotes

# String Indexing & Slicing

```python
p = "Rayan Chakraborty"
p[0:3] #remember that last index is not included in output
```

```python
# now let us look into negative indices
p[-5 : -1]
```

```python
p[-7:]
```

```python
# string[start : end : skip]
p[2:12:2]
```

```python
p[-10:-5:2]
```

# String Functions

```python
a = "Khushi"
```

```python
#len()
len(a)
```

```python
#string.endswith() - returns boolean output for 
a.endswith("hi")  #case sensitive
```

```python
#string.count() - gives number of occurences of a character
p.count("a")
```

```python
#string.capitalize()
c = "mama"
c.capitalize()
```

```python
#string.find(word) - used to find first occurence of a substring in a string
p.find("Chakr")
```

```python
#string.replace(old_word, new_word) - used to replace a substring with another substring 
c = "pancakes"
c.replace("a", "b")
```
