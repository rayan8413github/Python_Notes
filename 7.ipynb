# Functions in Python

# Basics

```python
def add(a,b):
    return a + b

add(9,4)
```

# Default arguments - default arguments always come after non default arguments

```python
def greet(name, mess="Hello"):
    return f"{mess}, {name}"

greet("Rayan")
greet("Rayan","Hi")
```

# Positional v/s Keyword arguments

```python
def info(name, age):
    print(name, age)

info("Ravi", 21)            # positional
info(age = 20, name = "Ramesh") # keyword (order doesn't matter here.)
```

# *args (variable number of positional arguments)

```python
def total(*args):
    print(args)
    return sum(args) #we made a sum function above we are using that only

total(4, 7, 2, 8)
```

# **kwargs (variable nummber of keywords) - gives answers in form of dictionaries

```python
def profile(**kwargs):
    print(kwargs)

profile(name = "Rayan", branch = "CME", institute = "NITA")
```

# If we combine everything in a function - order must be : - normal, default, *args, **kwargs

```python
def demo(a, b = 12, *args, **kwargs):
    print(a, args, b, kwargs)

demo(13, 65, 9, 5, 8, 0, age=21, rank=10)
```

# unpacking of *args and **kwargs

```python
n = [43, 61, 52]
d = {'a':10, 'b':30, 'c':90}
total(*n)
profile(**d)
```

# Lambda Function

```python
square = lambda x: x ** 2
square(4)   # 16

add = lambda a, b: a + b
add(7,4) #11
```
