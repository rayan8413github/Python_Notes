# Lists - Python lists are mutable containers in python to values of any datatypes

```python
# indexing is done in lists the same way as it is done in strings.
```

```python
L = [34, "RAM", 3.8, True]
# L[7] - out of index range hence it gives error
```

# list functions

```python
#list.append() - adds new element in last index
L.append("lalalalala")
L
```

```python
#L.copy() creates a new
a = [3, "aba", 2.0]
b = a.copy()
b
```

```python
#L.clear() - removes all elements of list and returns empty list
L1 = [3, 8, 9, 12, "k", 6]
L1.clear()
print(L1)
```

```python
# L.count() - gives number of instances of an element
L2 = [4, 7, 1, 4, 6, 6, 8, 3, 2, 1]
L2.count(1)
```

```python
# L1.extend(L2) - adds L2 at end of L1
a = [1, 2, 3]
b = ["a", "b", "c"]
a.extend(b)
a
```

```python
#L.index(value) - returns index of first occurence of value in parameter
L3 = [6, 10, 3, 4, 13, 6, 10, 21, 21]
L3.index(10)
```

```python
#L.insert(index ,value) - inserts given value at given index
L4 = [4.0, True, 71, 0, "hehe"]
L4.insert(2,"a")
L4
```

```python
# L.pop() - removes the last element from the list
L5 = ["gandhi", 90, 1.25, True, 1.01]
L5.pop()
L5
```

```python
# L.remove(element) - removes the first occurence of specified element from the list
d = [3, 7, 1, 5, 8, 9, 1]
d.remove(1)
d
```

```python
# reverse and sort - do the work as per their names
a = [7, 4, 2, 5, 1, 8, 9]
a.sort()
print(a)
b = [1, 5, 2, 4, 9, 3, 0, 2]
b.reverse()
print(b)
```
